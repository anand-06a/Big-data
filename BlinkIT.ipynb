```python
# import the pandas library
import pandas as pd
```

```python
# load the dataset
df=pd.read_excel('/content/BlinkIT Grocery Data.xlsx')
```

```python
# display 1st rows
df.head()
```

```python
# check the size of the data
df.shape
```

```python
# check column names
df.columns
```

```python
# check datatypes
df.dtypes
```

```python
# get dataset information (preprocessing)
df.info()
```

```python
# find missing values
df.isnull().sum()
```

```python
# handle missing values
# fill numarical missing values the mean.
df["Item Weight"]=df["Item Weight"].fillna(df["Item Weight"].mean())
```

```python
# fill categorical missing values using the mode.
df["Outlet Size"]=df["Outlet Size"].fillna(df["Outlet Size"].mode()[0])
```

```python
df.isnull().sum()
```

```python
# remove duplicate records
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
# summarizes numarical columns.
df.describe()
```

```python
# select specific columns
sales_data=df[["Item Type","Sales"]]
sales_data.head()
```

```python
# filter data
high_sales=df[df["Sales"]>200]
high_sales.head()
```

```python
# count the records
high_sales.shape
```

```python
# sort data
df.sort_values(by="Sales",ascending=False).head(10)
```

```python
# group data,total sales by item type
df.groupby("Item Type")["Sales"].sum()
```

```python
# average sales by outlet type
df.groupby("Outlet Type")["Sales"].mean()
```

```python
# number of products by outlet location
df.groupby("Outlet Location Type").size()
```

```python
# calculate multiple statistics together.
df.groupby("Item Type")["Sales"].agg(["count","sum","mean","max","min"])
```

```python
# create a new column
df["Sales Category"]=df["Sales"].apply(lambda x: "High" if x > 200 else "Low")
df[["Sales","Sales Category"]].head()
```

```python
# find top selling item types
top_items=df.groupby("Item Type")["Sales"].sum().sort_values(ascending=False)
top_items.head(10)
```

```python
# analyze outlet performance
outlet_sales = df.groupby("Outlet Identifier")["Sales"].sum().sort_values(ascending=False)
outlet_sales.head()
```

```python
# save the cleaned data
df.to_csv("BlinkIT_cleaned.csv",index=False)
```
